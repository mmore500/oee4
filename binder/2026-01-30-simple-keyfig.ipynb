# set up graphing utilities


```python
def letterscatter(*,data,x,y,cat):
    sns.scatterplot(
        data=data,
        x=x,
        y=y,
        color='k',
        edgecolor='k',
        s=200,
    )
    sns.scatterplot(
        data=data,
        x=x,
        y=y,
        hue=cat,
        edgecolor=None,
        s=120,
    )
    sns.scatterplot(
        data=data,
        x=x,
        y=y,
        color='w',
        style=cat,
        markers={
            phen : f'$\mathrm{{{phen}}}$'
            for phen in data[cat].unique()
        },
        legend=False,
    )
    plt.legend(
        bbox_to_anchor=(1.01, 1),
        borderaxespad=0,
        title='Morph',
        frameon=False,
    )
```

```python
def letterscatter_vline(*args, **kwargs):
    letterscatter(*args, **kwargs)
    plt.axvline(15,zorder=-2,c='k',ls=':')
    plt.axvline(45,zorder=-2,c='k',ls='--')
```

```python
# adapted from https://stackoverflow.com/a/49601444
def lighten_color(color, amount=0.5):
    """
    Lightens the given color by multiplying (1-luminosity) by the given amount.
    Input can be matplotlib color string, hex string, or RGB tuple.

    Examples:
    >> lighten_color('g', 0.3)
    >> lighten_color('#F034A3', 0.6)
    >> lighten_color((.3,.55,.1), 0.5)
    """
    import matplotlib.colors as mc
    import colorsys
    try:
        c = mc.cnames[color]
    except:
        c = color
    r, g, b = mc.to_rgb(c)
    return r + amount * (1 - r), g + amount * (1 - g), b + amount * (1 - b)
```

```python
from matplotlib.lines import Line2D


def lineage_decomposition(prev_stint):
    """Split the ancestry into lineages, latest-terminating first.

    Each leaf --- a stint that goes on to seed nothing --- is traced back
    through its ancestors until it meets a stint an earlier lineage has
    already taken, so every arrow falls in exactly one lineage and every
    lineage is one unbroken run. Working through the leaves latest first
    puts the line of descent reaching the final stint at the head of the
    list, and hangs each clade off the lineage it branched from.
    """
    seeds = {
        prev_stint[stint]
        for stint in range(len(prev_stint))
        if prev_stint[stint] != stint
    }
    leaves = [
        stint
        for stint in range(len(prev_stint))
        if stint not in seeds and prev_stint[stint] != stint
    ]

    taken, lineages = set(), []
    for leaf in sorted(leaves, reverse=True):
        chain, stint = [], leaf
        while prev_stint[stint] != stint and stint not in taken:
            chain.append((prev_stint[stint], stint))
            taken.add(stint)
            stint = prev_stint[stint]
        taken.add(stint)
        if chain:
            lineages.append(chain[::-1])
    return lineages


def lineage_drops(lineages, placement):
    """Where each branching stint drops, as {stint: (from row, to row)}."""
    row_of = {
        stint: row
        for chain, row in zip(lineages, placement)
        for _, stint in chain
    }
    drops = {}
    for chain, row in zip(lineages, placement):
        branch = chain[0][0]
        if branch in row_of:
            drops[branch] = (
                row_of[branch],
                max(row, drops.get(branch, (0, 0))[1]),
            )
    return drops


def layout_untangled(lineages, placement):
    """Does this row assignment keep every lineage and every drop clear?

    Three things have to hold: two lineages sharing a row may not run over
    each other, a lineage must sit below the one it branched off so that
    every branch turns downwards, and a drop may not cut across a lineage
    on a row it passes through.
    """
    by_row = {}
    for chain, row in zip(lineages, placement):
        by_row.setdefault(row, []).append((chain[0][0], chain[-1][1]))
    for runs in by_row.values():
        runs.sort()
        if any(nxt[0] < run[1] for run, nxt in zip(runs, runs[1:])):
            return False

    row_of = {
        stint: row
        for chain, row in zip(lineages, placement)
        for _, stint in chain
    }
    if any(
        row <= row_of[chain[0][0]]
        for chain, row in zip(lineages, placement)
        if chain[0][0] in row_of
    ):
        return False

    return not any(
        lo < branch < hi
        for branch, (top, bottom) in lineage_drops(lineages, placement).items()
        for row in range(top + 1, bottom)
        for lo, hi in by_row.get(row, ())
    )


def stack_lineages(lineages):
    """Lay out the lineages in rows, with a drop onto each branch.

    A whole lineage goes down at once, so it reads as a single run rather
    than as arrows scattered over several rows. Each one takes the highest
    row that leaves the layout untangled; where no row does, a fresh one is
    opened between the branch and whatever is in the way. A second pass
    then lifts every lineage as far as it will go, so the stack ends up as
    shallow as the constraints allow.

    Returns the row per arrow, the number of rows, and the drops.
    """
    placement = []
    for index in range(len(lineages)):
        grown = lineages[: index + 1]
        row = next(
            (
                row
                for row in range(max(placement, default=-1) + 2)
                if layout_untangled(grown, placement + [row])
            ),
            None,
        )
        if row is None:
            row, placement = next(
                (opened, shifted)
                for opened in range(max(placement, default=-1) + 2)
                for shifted in [[r + (r >= opened) for r in placement]]
                if layout_untangled(grown, shifted + [opened])
            )
        placement.append(row)

    lifted = True
    while lifted:
        lifted = False
        for index in sorted(range(len(placement)), key=placement.__getitem__):
            for row in range(placement[index]):
                raised = placement[:index] + [row] + placement[index + 1 :]
                if layout_untangled(lineages, raised):
                    placement, lifted = raised, True
                    break

    closed = {row: r for r, row in enumerate(sorted(set(placement)))}
    placement = [closed[row] for row in placement]
    assert layout_untangled(lineages, placement), "tangled layout"

    return (
        {
            arrow: row
            for chain, row in zip(lineages, placement)
            for arrow in chain
        },
        len(closed),
        lineage_drops(lineages, placement),
    )


def stacked_ancestry_arrows(
    *, fig, ax1, prev_stint, band=0.7, lw=0.4, marker_size=2.0
):
    """Draw the ancestry as level runs layered into the morph strip.

    Each arrow runs horizontally from its ancestor's key to its
    descendant's, ending in a dot where the descendant goes on to seed a
    stint of its own and in a cross where the lineage stops there; a
    vertical drops from each branching dot to the lineages leaving it.
    The line of descent reaching the final stint takes the top row, on
    the top edge of the strip, and every other lineage hangs below the
    one it left. The stack takes up `band` of the strip's height whatever
    its depth.
    """
    lineages = lineage_decomposition(prev_stint)
    assignment, n_rows, drops = stack_lineages(lineages)
    assert len(assignment) == sum(
        prev_stint[stint] != stint for stint in range(len(prev_stint))
    ), "every arrow belongs to exactly one lineage"

    carries_on = {to for to, _ in assignment}
    into_fig = fig.transFigure.inverted()

    def key(stint, y):
        iax = min(stint // 10, len(ax1) - 1)
        return into_fig.transform(
            ax1[iax].transData.transform((stint - iax * 10 + 0.5, y)),
        )

    y_top, y_bottom = key(0, 0)[1], key(0, 1)[1]
    step = band * (y_top - y_bottom) / max(n_rows - 1, 1)

    def at(row):
        return y_top - row * step

    for stint, (top, bottom) in drops.items():
        x = key(stint, 0)[0]
        fig.add_artist(
            Line2D(
                [x, x],
                [at(top), at(bottom)],
                transform=fig.transFigure,
                color="black",
                lw=lw,
                solid_capstyle="butt",
                zorder=4,
            )
        )

    for (to, fr), row in assignment.items():
        fig.add_artist(
            Line2D(
                [key(to, 0)[0], key(fr, 0)[0]],
                [at(row), at(row)],
                transform=fig.transFigure,
                color="black",
                lw=lw,
                solid_capstyle="butt",
                marker="o" if fr in carries_on else "x",
                markersize=marker_size,
                markeredgewidth=lw * 1.5,
                markevery=[1],
                zorder=5,
            )
        )
```
